# Исправление неверного порядка ячеек

```python
print(project_title)
project_title = "Text Lab"
project_title.upper()
```

код не работает, так как ячейки расположены в неправильном порядке.
print не запускается, так как мы еще не задали переменную project_title

```python
project_title = "Text Lab"
print(project_title)
project_title.upper()
```

поменяли местами 1 и 2 ячейки, код заработал
